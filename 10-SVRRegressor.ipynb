```python
X = df.drop("price", axis=1)
y = df["price"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.25, random_state=15)
```

```python
from sklearn.preprocessing import LabelEncoder # label encoding sirali verilerde kullanilabilir
                                               #(sirali olmayan verilerde kullanilirsa linear modellerde sorun cikarir.)

label_encoder = LabelEncoder()

for col in ["cut", "color", "clarity"]:
    X_train[col] = label_encoder.fit_transform(X_train[col])
    X_test[col] = label_encoder.transform(X_test[col])
```

```python
X_train.head()
```

```python
X_train["cut"].value_counts()
```

```python
X_test["cut"].value_counts()
```

```python
X_train.info()
```

```python
from sklearn.preprocessing import StandardScaler # RobustScaler -> outlier degerli bir veri setinde kullanilabilir.

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

linear = LinearRegression()
linear.fit(X_train_scaled, y_train)
y_pred = linear.predict(X_test_scaled)
```

```python
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

plt.figure(figsize=(12,6))
sns.set_theme(style=('whitegrid'))
sns.scatterplot(
    x = y_test,
    y = y_pred,
    legend = False,
    hue = y_test,
    alpha = 0.85,
    palette = "viridis"
)
plt.suptitle("Lineer Regression - Actual vs Predicted",fontsize = 12)

plt.title(f'R2 Score: {r2:.4f} | MSE: {mse:.4f} | MAE: {mae:.4f}', fontsize=10, loc='left', style='italic')
plt.xlabel('Predicted Values')
plt.ylabel('Actual Values')

plt.plot([y_test.min(),y_test.max()], [y_test.min(), y_test.max()], linewidth=1.5, alpha=0.8, color = 'r')

plt.show()
```

```python
from sklearn.svm import SVR
```

```python
svr = SVR()
svr.fit(X_train_scaled, y_train)
y_pred = svr.predict(X_test_scaled)

mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

plt.figure(figsize=(12,6))
sns.set_theme(style=('whitegrid'))
sns.scatterplot(
    x = y_test,
    y = y_pred,
    legend = False,
    hue = y_test,
    alpha = 0.85,
    palette = "viridis"
)
plt.suptitle("Support Vektor Regression - Actual vs Predicted",fontsize = 12)

plt.title(f'R2 Score: {r2:.4f} | MSE: {mse:.4f} | MAE: {mae:.4f}', fontsize=10, loc='left', style='italic')
plt.xlabel('Predicted Values')
plt.ylabel('Actual Values')

plt.plot([y_test.min(),y_test.max()], [y_test.min(), y_test.max()], linewidth=1.5, alpha=0.8, color = 'r')

plt.show()
```
