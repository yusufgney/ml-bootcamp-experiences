```python
def correlation_for_dropping(df, threshold):
    columns_to_drop = set()
    corr = df.corr()
    for i in range(len(corr.columns)):
        for j in range(i):
            if abs(corr.iloc[i, j]) > threshold:
                columns_to_drop.add(corr.columns[i])
    return columns_to_drop   
```

```python
columns_dropping = correlation_for_dropping(X_train, 0.85)
```

```python
columns_dropping
```

```python
X_train.drop(columns_dropping, axis = 1, inplace = True)
X_test.drop(columns_dropping, axis = 1, inplace = True)
```

```python
X_test.shape
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
plt.subplots(figsize = (20,10))
plt.subplot(1,2,1)
sns.boxplot(data = X_train)
plt.title('X_train')
plt.subplot(1,2,2)
sns.boxplot(data = X_train_scaled)
plt.title('X_train_scaled')
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
def adjusted_r2_score(r2, y_test, X_test):
    return 1 - (1-r2)*(len(y_test)-1)/(len(y_test)-X_test.shape[1]-1)
```

```python
linear = LinearRegression()
linear.fit(X_train_scaled, y_train)
y_pred = linear.predict(X_test_scaled)
```

```python
mae = mean_absolute_error(y_test, y_pred)  
mse = mean_squared_error(y_test, y_pred)   
r2 = r2_score(y_test, y_pred)        
adjust_r2 = adjusted_r2_score(r2, y_pred, X_test_scaled)

metrics_score = {'mse: ':mse, 'mae: ':mae, 'r2_score: ':r2, 'adjusted_r2_score: ': adjust_r2}
for key, value in metrics_score.items():
    print(f"{key}{value}")
```

```python
plt.figure(figsize = (10,5))
sns.set_theme(style = ('whitegrid'))
plt.scatter(y_test, y_pred)
plt.title('Linear')
plt.show()
```
