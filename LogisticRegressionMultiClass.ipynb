```python
import numpy as np 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv("7-cyber_attack_data.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
df.describe()
```

```python
X=df.drop("attack_type", axis=1)
y=df["attack_type"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.3, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
score = accuracy_score(y_pred, y_test)
print("score : ", score, "\n")
print(classification_report(y_pred, y_test))
print("confusion matrix : \n", confusion_matrix(y_pred, y_test))
```

```python
penalty = ["l1", "l2", "elasticnet"]
c_values = [100, 10, 1, 0.1, 0.01]
solver = ["newton-cg", "newton-cholesky", "lbfgs", "liblinear", "sag", "saga"]
```

```python
params = dict(penalty=penalty, C=c_values, solver=solver)
```

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold
```

```python
cv = StratifiedKFold()
```

```python
grid = GridSearchCV(estimator=model, param_grid=params, cv=cv, scoring="accuracy", n_jobs=-1)
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred1 = grid.predict(X_test)
```

```python
score = accuracy_score(y_pred1, y_test)
print("score : ", score, "\n")
print(classification_report(y_pred1, y_test))
print("confusion matrix : \n", confusion_matrix(y_pred1, y_test))
```

```python
# one vs rest
```

```python
from sklearn.multiclass import OneVsOneClassifier, OneVsRestClassifier
```

```python
onevsonemodel = OneVsOneClassifier(LogisticRegression())
```

```python
onevsrestmodel = OneVsRestClassifier(LogisticRegression())
```

```python
onevsonemodel.fit(X_train, y_train)
y_pred2 = onevsonemodel.predict(X_test)

score = accuracy_score(y_pred2, y_test)
print("score : ", score, "\n")
print(classification_report(y_pred2, y_test))
print("confusion matrix : \n", confusion_matrix(y_pred2, y_test))
```

```python
onevsrestmodel.fit(X_train, y_train)
y_pred3 = onevsonemodel.predict(X_test)

score = accuracy_score(y_pred3, y_test)
print("score : ", score, "\n")
print(classification_report(y_pred3, y_test))
print("confusion matrix : \n", confusion_matrix(y_pred3, y_test))
```
