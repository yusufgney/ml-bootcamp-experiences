```python
from sklearn.ensemble import IsolationForest
```

```python
df["Class"].value_counts(normalize=True)
```

```python
fraud_ratio = 0.001727
```

```python
iso = IsolationForest(
    n_estimators=200, 
    contamination=fraud_ratio,
    max_features=1.0, # featureslari secer 1.0 yaparsak butun feature'lari alir.
    max_samples="auto",
    random_state=42,
    n_jobs=-1
)# contamination -> dataset icerisindeki outlierlarin orani. 
```

```python
X = df.drop("Class", axis=1)
```

```python
iso.fit(X)
```

```python
df["iso_raw"] = iso.predict(X)
```

```python
df["iso_raw"].unique()
```

```python
# 1 -> normal
# -1 -> anomaly
```

```python
df["iso_pred"] = df["iso_raw"].map({1 : 0, -1 : 1})
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop(["Class", "iso_raw", "iso_pred"], axis=1)
y = df["Class"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

```python
iso_test_pred = df.loc[X_test.index, "iso_pred"]
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, iso_test_pred))
```

```python
df["Anomaly_score"] = iso.decision_function(X)
```

```python
sns.kdeplot(df[df["Class"] == 0]["Anomaly_score"], label="Normal")
sns.kdeplot(df[df["Class"] == 1]["Anomaly_score"], label="Fraud")
plt.title("Anomaly Score Distibution")
plt.legend()
plt.show()
```

```python
df["Anomaly_score"].quantile(0.01) # top %1 i gormeye calisiyoruz.
```

```python
threshold = df["Anomaly_score"].quantile(0.01)
```

```python
df["iso_custom"] = (df["Anomaly_score"] < threshold).astype(int)
```

```python
print(classification_report(df["Class"], df["iso_custom"]))
```

```python
sns.kdeplot(df[df["Class"] == 0]["Anomaly_score"], label="Normal")
sns.kdeplot(df[df["Class"] == 1]["Anomaly_score"], label="Fraud")
plt.axvline(threshold, c='r', label="Decision Threshold")
plt.title("Anomaly Score Distibution")
plt.legend()
plt.show()
```

```python
df_sorted = df.sort_values("Anomaly_score")
```

```python
k = int(len(df_sorted) * 0.01)
top_k = df_sorted.head(k)
```

```python
fraud_captured = top_k["Class"].sum()
total_frauds = df["Class"].sum()

recall_at_k = f"%{(fraud_captured / total_frauds) * 100:.4f}"
```

```python
recall_at_k
```

```python
percentage = np.linspace(0.001, 0.05, 30)
recall = []

total_frauds = df["Class"].sum()

for p in percentage:
    k = int(len(df_sorted) * p)
    captured = df_sorted.head(k)["Class"].sum()
    recall.append(captured / total_frauds)
```

```python
plt.plot(percentage * 100, recall)
plt.show()
```
