```python
# independent and dependent features
X = df.iloc[:,:-1] # son kolon haric hepsini al.
y = df.iloc[:,-1] # sadece son kolonu al.

# X = df[['Study Hours', 'Sleep Hours', 'Attendance Rate', 'Social Media Hours']]
# y = df['Exam SCore']
```

```python
X.tail()
```

```python
# train - test split
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.25, random_state = 15)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
from sklearn.linear_model import LinearRegression
```

```python
regression = LinearRegression()
```

```python
regression.fit(X_train, y_train)
```

```python
X_test
```

```python
new_student = [[5, 7, 90, 1]]
```

```python
new_student_scaled = scaler.transform(new_student) 
```

```python
regression.predict(new_student_scaled)
```

```python
def adjusted_r2_score(r2, y_test, X_test):
    return 1 - (1-r2)*(len(y_test)-1)/(len(y_test)-X_test.shape[1]-1)
```

```python
# prediction
```

```python
y_pred = regression.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
adjust_r2 = adjusted_r2_score(r2, y_test, X_test)

metrics_score = {'mse: ':mse, 'mae: ':mae, 'r2_score: ':r2, 'adjusted_r2_score: ': adjust_r2}
for key, value in metrics_score.items():
    print(f"{key}{value}")
```
